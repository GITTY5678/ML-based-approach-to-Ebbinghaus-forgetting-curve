```python
import pandas as pd
import matplotlib.pyplot as plt

data=pd.read_csv("/home/harihara-suthan-s/ML-based-approach-to-Ebbinghaus-forgetting-curve/training/ebbinghaus_combined_2600_rows_cleaned_2.csv")
data.head(5)
```

```python
data["days"]=data["revision_timing"]/24
data["days_round"]=data["days"].round()
avg_curve=(
    data.groupby("days_round")["retention"].mean().reset_index()
)
plt.plot(avg_curve["days_round"],avg_curve["retention"],marker="o")
plt.xlabel("days")
plt.ylabel("retention")
plt.grid("True")
plt.show()
```
