```python
import numpy as np
import pandas as pd
```

```python
n=3000
np.random.seed(42)

rows=[]

session_bonus_map = {
    "morning": 5,
    "afternoon": 2,
    "evening": 0,
    "night": -3
}
session_choices=[
    "morning",
    "afternoon",
    "evening",  
    "night"
]

for _ in range(n):
    revision_timing = np.clip(
        np.random.exponential(scale=120),
        1,
        720
    )

    nth_revision = np.random.randint(0, 10)

    sleep_duration = np.random.uniform(3, 10)

    topic_difficulty = np.random.randint(1, 11)

    study_duration = np.random.uniform(10, 180)

    session_time = np.random.choice(session_choices)

    session_bonus = session_bonus_map[session_time]

    # -------------------------
    # Memory Stability (S)
    # -------------------------

    base_strength = 40

    revision_bonus = 25 * np.log1p(nth_revision)

    sleep_bonus = 4 * (sleep_duration - 6)

    difficulty_penalty = 3 * topic_difficulty

    study_bonus = 10 * np.log1p(
        study_duration / 30
    )

    S = (
        base_strength
        + revision_bonus
        + sleep_bonus
        - difficulty_penalty
        + study_bonus
        + session_bonus
    )

    # Safety check
    S = max(S, 5)

    # -------------------------
    # Forgetting Curve
    # -------------------------

    retention = (
        100 *
        np.exp(-revision_timing / S)
    )

    # Human variability
    noise = np.random.normal(
        loc=0,
        scale=5
    )

    retention += noise

    retention = np.clip(
        retention,
        0,
        100
    )

    # -------------------------
    # Save Row
    # --------------------------    
    rows.append({
        "revision_timing": revision_timing,
        "nth_revision": nth_revision,
        "sleep_duration": sleep_duration,
        "topic_difficulty": topic_difficulty,
        "study_duration": study_duration,
        "session_time": session_time,
        "retention": retention
    })
df=pd.DataFrame(rows)
df.to_csv("synthetic_forgetting_curve_data.csv", index=False)
```
