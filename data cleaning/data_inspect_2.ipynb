```python
import numpy as np
import pandas as  pd
```

```python
data=pd.read_csv("/home/harihara-suthan-s/ML-based-approach-to-Ebbinghaus-forgetting-curve/dataset creation/synthetic_forgetting_curve_data.csv")
data.describe()
```

```python
data.describe(include='all')
```

```python
time_mapping = {
    "morning": 0,
    "afternoon": 1,
    "evening": 2,
    "night": 3
}

data["session_time"] = data["session_time"].map(time_mapping)
```

```python
data.describe(include='all')
```

```python
data.head(5)
```

```python
import matplotlib.pyplot as plt

plt.scatter(data["revision_timing"], data["retention"])
plt.xlabel("Revision Timing")
plt.ylabel("Retention Percentage")
plt.title("Revision Timing vs Retention Percentage")   
plt.show()
```

```python
data.boxplot(
    column="retention",
    by="nth_revision",
    grid=False
)

plt.title("Retention Percentage vs Nth Revision")
plt.suptitle("")  # removes default pandas subtitle
plt.xlabel("Nth Revision")
plt.ylabel("Retention Percentage (%)")

plt.show()
```

```python
plt.scatter(data["sleep_duration"], data["retention"])
plt.xlabel("Sleep Duration")
plt.ylabel("Retention Percentage")
plt.title("Sleep Duration vs Retention Percentage")
plt.show()
```

```python
print(data["sleep_duration"].corr(data["retention"]))
```
